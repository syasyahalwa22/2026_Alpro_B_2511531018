## 2.1 Dasar Teori
Perulangan (looping) adalah struktur kontrol yang menjalankan sekumpulan instruksi secara berulang sehingga programmer tidak perlu menulis perintah yang sama berkali-kali. Pada Python, perulangan for dipakai ketika jumlah pengulangan sudah diketahui, dengan mengambil nilai satu per satu dari suatu urutan, biasanya dibuat melalui fungsi range(). Fungsi range(stop) menghasilkan bilangan dari 0 sampai stop - 1, sedangkan range(start, stop) dimulai dari start. Nilai stop tidak ikut dihasilkan, sehingga untuk mencetak 1 sampai n digunakan range(1, n + 1). Blok kode yang diulang ditandai dengan indentasi, sehingga kesalahan indentasi dapat mengubah hasil program.

Dalam program, masukan dari pengguna diterima dengan input() yang menghasilkan string, lalu diubah menjadi bilangan bulat dengan int(). Hasilnya ditampilkan dengan print(), di mana parameter end=" " membuat cetakan tetap berada di baris yang sama, dan print() kosong dipakai untuk pindah baris. Nilai variabel dapat disisipkan ke dalam teks menggunakan f-string.

Selain itu, program memakai percabangan if-elif-else untuk memilih tindakan berdasarkan kondisi, dengan operator modulus (%) untuk menentukan bilangan genap atau ganjil. Untuk menghitung total, digunakan akumulator, yaitu variabel yang diberi nilai awal 0 lalu diperbarui di setiap putaran dengan pola jumlah = jumlah + i. Terakhir, perulangan bersarang (nested loop) adalah perulangan di dalam perulangan, di mana loop dalam dijalankan penuh setiap kali loop luar berputar satu kali. Pada pencetakan pola, loop luar umumnya mengatur baris dan loop dalam mengatur kolom.

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2511531018"
Mencetak tulisan "Perulangan ke-..." sebanyak jumlah yang dimasukkan user, dengan penomoran mulai dari 0. Kalau input 3, hasilnya "Perulangan ke-0", "ke-1", dan "ke-2".

```python
ulang_1018 = int(input("Masukkan jumlah perulangan: "))
for i_1018 in range(ulang_1018):
    print(f"Perulangan ke-{i_1018}")
```

**Penjelasan :**
1. Menampilkan teks dan menerima masukan dari user.
2. Mengubah string tersebut menjadi bilangan bulat.
3. Menyimpan jumlah perulangan ke variabel.
4. Mengulang dengan i_1018 bernilai 0 sampai ulang_1018 - 1.
5. Mencetak teks dengan nilai i_1018 yang dissipkan lewat f-string.

### 2.2.2 Program "perulangan_for2_2511531018"

Menampilkan dua deret angka dari jumlah yang sama: satu deret mulai dari 0 sampai n-1, dan satu deret mulai dari 1 sampai n. Program ini dipakai untuk menunjukkan perbedaan cara kerja range().

```python
ulang_1018 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_1018-1)
for i_1018 in range(ulang_1018):
    print(i_1018, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_1018)
for i_1018 in range(1, ulang_1018+1):
    print(i_1018, end=" ")
```

**Penjelasan :**
1. Mencetak keterangan rentang, dengan batas akhir ulang - 1.
2. range(n) menghasilkan 0, 1, ..., n-1.
3. Mencetak angka diikuti spasi (bukan baris baru) karena end=" ".
4. Mencetak baris kosong untuk pindah baris.
5. Menghasilkan 1, 2, ..., n. Batas akhir tidak ikut, maka ditambah 1.

### 2.2.3 Program "perulangan_for3_2511531018"
Menampilkan deret penjumlahan dari 1 sampai n beserta hasil totalnya, misalnya 1 + 2 + 3 = 6. (Pada kode aslinya hasil tercetak terpecah karena print() menjorok di dalam loop, jadi perlu digeser ke luar loop.)

```python
ulang_1018 = int(input("Masukkan jumlah perulangan: "))

jumlah_1018 = 0
for i_1018 in range(1, ulang_1018 + 1):
    print(i_1018, end=" ")
    jumlah_1018 = jumlah_1018 + i_1018

    if i_1018 < ulang_1018:
        print("+", end=" ")
    else:
        print("=", jumlah_1018, end=" ")
print()
print("Jumlah =", jumlah_1018)
```

**Penjelasan :**
1. Variabel penampung total (akumulator), awalnya 0.
2. Mengulang dari 1 sampai n.
3. Mencetak angka saat ini tanpa pindah baris.
4. Menambahkan angka saat ini ke total, jika belum angka terakhir, cetak tanda +
5. Jika sudah angka terakhir, cetak = dan hasil totalnya. Pindah baris, lalu cetak total.

### 2.2.4 Program "jumlah_genap_2511531018"
Menampilkan semua bilangan genap dari 1 sampai batas yang dimasukkan, lalu menjumlahkannya. Contoh input 6: 2 + 4 + 6 = 12.

```python
ulang_1018 = int(input("Masukkan nilai batas: "))

jumlah_1018 = 0
for i_1018 in range(1, ulang_1018 + 1):
    if i_1018 % 2 == 0:
        print(i_1018, end=" ")
        jumlah_1018 = jumlah_1018 + i_1018

        if i_1018 < ulang_1018:
            print("+", end=" ")
        else:
            print("=", jumlah_1018, end=" ")
print()
print("Jumlah =", jumlah_1018)
```

**Penjelasan :**
1. Akumulator total. 
2. Mengulang dari 1 sampai batas. % adalah sisa bagi. Jika sisa bagi 2 sama dengan 0, bilangan itu genap.
3. Mencetak bilangan genap dan menambahkannya ke total.
4. Mencetak + atau = seperti pada program 3.
5. Di luar loop, sehingga hanya dijalankan sekali di akhir.

### 2.2.5 Program "nested_for1_2511531018"
Mencetak segitiga angka rata kanan: tiap baris berisi titik di kiri dan nomor baris di ujung kanan, dengan jumlah titik yang makin berkurang.

```python
batas_1018 = int(input("Masukkan nilai batas: "))
for line in range(1, batas_1018 + 1):
    for i_1018 in range(1, (-1 * line + batas_1018) + 1):
        print(".", end=" ")
    print(line)
```

**Penjelasan :**
1. Loop luar, mewakili baris ke-1 sampai ke-batas.
2. Loop dalam berjalan sebanyak batas - line kali. Semakin besar nomor baris, semakin sedikit titiknya.
3. Mencetak titik tanpa pindah baris dan setelah titik selesai, cetak nomor baris dan pindah baris. 

### 2.2.6 Program "nested_for2_2511531018"
Mencetak persegi bintang berukuran n × n, yaitu n baris dengan n bintang di tiap barisnya.

```python
batas_1018 = int(input("Masukkan nilai batas: "))
for i_1018 in range(1, batas_1018 + 1):
    for j_1018 in range(1, batas_1018 + 1):
        print("*", end=" ")
    print() #pindah ke baris berikutnya
```

**Penjelasan :**
1. Loop luar mengatur jumlah baris dan loop dalam mengatur jumlah kolom di tiap baris.
2. Mencetak bintang di satu baris yang sama.
3. Pindah baris setelah satu baris penuh selesai.

### 2.2.7 Program "nested_for3_2511531018"
Membuat tabel penjumlahan dari 0 sampai n, di mana setiap kotak berisi hasil penjumlahan nomor baris dan nomor kolomnya.

```python
batas_1018 = int(input("Masukkan nilai batas: "))
for i_1018 in range(batas_1018 + 1):
    for j_1018 in range(batas_1018 + 1):
        print(i_1018 + j_1018, end=" ")
    print() #pindah ke baris berikutnya
```

**Penjelasan :**
1. Menghasilkan 0 sampai batas (dimulai dari 0, bukan 1). Jadi tabelnya berukuran (batas+1) × (batas+1).
2. Nilai baris dan nilai kolom.
3. Mencetak hasil penjumlahan baris dan kolom pada sel tersebut dan pindah baris.

### 2.2.8 Program "nested_for4_2511531018"
Menggambar bingkai persegi panjang (sudut *, garis atas-bawah =, sisi |) yang di dalamnya berisi pola titik yang melebar dari tengah lalu menyempit lagi, simetris atas-bawah. Tinggi pola harus bilangan genap, kalau ganjil muncul pesan error.

```python
tinggi_1018 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))
if tinggi_1018 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1018 = tinggi_1018
    c_1018 = a_1018
    lebar_1018 = (2 * tinggi_1018) - 2

    for i_1018 in range(1, tinggi_1018 + 1):
        b_1018 = c_1018 + 1

        for j_1018 in range(1, lebar_1018 + 1):
            if i_1018 == 1 or i_1018 == tinggi_1018: 
                if j_1018 == 1 or j_1018 == lebar_1018:
                    print("*", end=" ")
                else:
                    print("=", end=" ")
            else:
                if j_1018 == 1 or j_1018 == lebar_1018:
                    print("|", end=" ")
                else:
                    if j_1018 == c_1018: 
                        print("<", end=" ")
                    elif j_1018 == b_1018:
                        print(">", end=" ")
                    elif j_1018 == (lebar_1018 - c_1018):
                        print("<", end=" ")
                    elif j_1018 == (lebar_1018 - c_1018 + 1):
                        print(">", end=" ")
                    elif j_1018 > b_1018 and j_1018 < (lebar_1018 - c_1018):
                        print(".", end=" ")
                    else:
                        print(" ", end=" ")
        print()
        a_1018 -= 2

        if a_1018 <= 0:
            c_1018 = (-a_1018) + 2
        else:
            c_1018 = a_1018
                    
```

**Penjelasan :**
1. Meminta tinggi pola dari user, lalu mengubahnya menjadi bilangan bulat.
2. Jika sisa bagi 2 bukan 0 (ganjil), tampilkan pesan error dan program berhenti.
3. Variabel pengatur yang nilainya turun 2 di setiap baris.
4. Posisi kolom tempat < pertama dicetak pada baris tersebut.
5. Jumlah kolom. Untuk tinggi 10, lebarnya 18.

### 2.2.9 Program "latihan_pekan5_2511531018"
Mencetak piramida bintang sesuai tinggi yang dimasukkan: jumlah bintang bertambah 2 di tiap baris (1, 3, 5, ...) dan diratakan ke tengah dengan spasi di kiri. (Pada kode aslinya print() terakhir perlu digeser ke dalam loop luar agar tiap baris pindah ke baris baru.)

```python
tinggi_1018 = int(input("Masukkan tinggi segitiga: "))

for i_1018 in range(1, tinggi_1018 + 1):
    for j_1018 in range(tinggi_1018 - i_1018):
        print(" ", end=" ")

    for j_1018 in range(2 * i_1018 - 1):
        print("*", end=" ")
    print() 
```

**Penjelasan :**
1. Input meminta user memasukkan tinggi segitiga, mengubah masukan menjadi bilangan bulat, dan menyimpan jumlah baris segitiga.
2. Loop luar untuk mengatur, i_1018 adalah nomor baris, dari 1 sampai tinggi.
3. Loop dalam pertama (spasi kiri) berjalan ebanayak tinggi - i kali, mencetak spasi sebagai identitas agar bintang bergeser ke tengah, semakin besar nomor baris, semakin sedikit spasinya.
4. Loop dalam kedua (bintang) berjalan sebanyak 2*i - 1 kali, lalu mencetak bintang berjumlah ganjil yang bertambah 2 tiap baris.

<em> Studi Kasus :  Pola Jam Pasir Kristal Palindromik Berbingkai (Framed Palindromic Crystal Hourglass)
Mahasiswa diminta merancang program konsol interaktif untuk mencetak pola bangun geometri dua dimensi berupa Jam Pasir Kristal Palindromik yang terbungkus rapi di dalam bingkai berpagar. Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.
Program menerima masukan nilai integer positif N dari pengguna yang mewakili ukuran skala jam pasir (skala angka tertinggi pada puncak dan alas).

Struktur Pembentukan Pola (3 Fase + 2 Bingkai Horizontal):

1. Bingkai Pembatas Horizontal (Batas Atas dan Bawah):

○ Karakter pagar sudut (#) di ujung kiri dan kanan.

○ Karakter garis ganda (=) di bagian tengah sebanyak (4*N+5) karakter.

2. Fase 1: Jam Pasir Atas (Reduksi Angka Menurun: Baris N turun sd 1):

○ Sisi kiri dibatasi garis tegak (|) dan satu spasi padding.

○ Spasi penyeimbang kiri sebanyak 2*(N - baris).

○ Deret angka mundur dari baris turun ke 1, dipisahkan dengan spasi (contoh baris 4: "4 3 2 1 ").

○ Poros kristal tengah: karakter "<*>".

○ Deret angka maju dari 1 naik ke baris, diawali spasi (contoh baris 4: " 1 2 3 4").

○ Spasi penyeimbang kanan sebanyak 2* (N - baris).

○ Satu spasi padding dan garis tegak (|) di sisi kanan.

3. Fase 2: Poros Titik Pusat Jam Pasir (Titik Nol / Singularitas):

○ Karakter garis tegak pembatas (|) di kiri.

○ Spasi penyeimbang kiri sebanyak (2*N+1).

○ Karakter poros kristal tunggal "<*>".

○ Spasi penyeimbang kanan sebanyak (2*N+1).

○ Karakter garis tegak pembatas (|) di kanan.

4. Fase 3: Jam Pasir Bawah (Ekspansi Angka Menaik: Baris 1 naik sd N):

○ Sisi kiri dibatasi garis tegak (|) dan satu spasi padding.

○ Spasi penyeimbang kiri sebanyak 2*(N - baris).

○ Deret angka mundur dari baris turun ke 1, dipisahkan dengan spasi.

○ Poros kristal tengah: karakter "<*>".

○ Deret angka maju dari 1 naik ke baris, diawali spasi.

○ Spasi penyeimbang kanan sebanyak 2* (N - baris).

○ Satu spasi padding dan garis tegak (|) di sisi kanan.

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_1018 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Bingkai Atas 
print("#", end="")
for k_1018 in range(4 * n_1018 + 5):
    print("=", end="")
print("#", end="")
print()
 
# Fase 1: Jam Pasir Atas (baris N turun sampai 1) 
for baris_1018 in range(n_1018, 0, -1):
    print("|", end="")
    print(" ", end="")                                  

    for spasi_1018 in range(2 * (n_1018 - baris_1018)): 
        print(" ", end="")

    for angka_1018 in range(baris_1018, 0, -1):         
        print(angka_1018, end=" ")

    print("<*>", end="")                               

    for angka_1018 in range(1, baris_1018 + 1):         
        print(" ", end="")
        print(angka_1018, end="")

    for spasi_1018 in range(2 * (n_1018 - baris_1018)): 
        print(" ", end="")

    print(" ", end="")                                  
    print("|", end="")
    print()

# Fase 2: Poros Titik Pusat 
print("|", end="")
for spasi_1018 in range(2 * n_1018 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_1018 in range(2 * n_1018 + 1):
    print(" ", end="")
print("|", end="")
print()

# Fase 3: Jam Pasir Bawah (baris 1 naik sampai N) 
for baris_1018 in range(1, n_1018 + 1):
    print("|", end="")
    print(" ", end="")

    for spasi_1018 in range(2 * (n_1018 - baris_1018)):
        print(" ", end="")

    for angka_1018 in range(baris_1018, 0, -1):
        print(angka_1018, end=" ")

    print("<*>", end="")

    for angka_1018 in range(1, baris_1018 + 1):
        print(" ", end="")
        print(angka_1018, end="")

    for spasi_1018 in range(2 * (n_1018 - baris_1018)):
        print(" ", end="")

    print(" ", end="")
    print("|", end="")
    print()

# Bingkai Bawah 
print("#", end="")
for k_1018 in range(4 * n_1018 + 5):
    print("=", end="")
print("#", end="")
print()
```

**Penjelasan :**
1. Program terdiri dari lima bagian yaitu bingkai atas, tiga fase pola, dan bingkai bawah. Seluruh karakter dicetak dengan perulangan dan end="", tanpa perkalian string, sesuai ketentuan.
2. Bingkai atas dan bawah diawali #, lalu loop mencetak = sebanyak 4*N+5 kali, kemudian ditutup #.

Setiap baris dicetak berurutan: | dan satu spasi padding, spasi penyeimbang kiri sebanyak 2*(N - baris), angka mundur dari baris ke 1 (masing-masing diikuti spasi), poros <*>, angka maju dari 1 ke baris (masing-masing didahului spasi), spasi penyeimbang kanan, satu spasi padding, lalu |.
4. Fase 2 hanya mencetak satu baris pusat: |, 2*N+1 spasi, <*>, 2*N+1 spasi, dan |.
5. Fase 3 memakai isi baris yang sama dengan Fase 1, tetapi baris naik dari 1 ke N lewat range(1, n + 1).
6. Lebar isi setiap baris selalu 4*N+5 karakter, sama dengan lebar = pada bingkai, sehingga bentuknya rapi. Pola ini rata jika dilihat dengan font monospace. Untuk N ≥ 10, angka dua digit akan menggeser kerataan karena lebarnya lebih dari satu karakter.
