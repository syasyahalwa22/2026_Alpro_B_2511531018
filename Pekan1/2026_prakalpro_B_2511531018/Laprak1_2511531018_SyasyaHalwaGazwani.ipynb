## 2.1 Dasar Teori
Pemrograman adalah cara menulis yang unik. Dengan menggunakan bahasa tertentu, dimungkinkan untuk menginstruksikan mesin untuk menjalankan beberapa tindakan. Proses ini memiliki banyak analogi dengan koreografi dan penulisan teater. Bahasa pemrograman adalah tool yang wajib dikuasai oleh para programmer. Bahasa program atau pemrograman sendiri adalah sekumpulan instruksi yang diberikan kepada komputer agar bisa melakukan tugas-tugas tertentu dalam menyelesaikan sebuah masalah. Ada banyak bahasa pemrograman, salah satunya adalah Python. 

Python merupakan bahasa pemrograman komputer yang biasa dipakai untuk membangun situs, software/aplikasi, mengotomatiskan tugas dan melakukan analisis data. Bahasa pemrograman ini termasuk bahasa tujuan umum. Artinya, ia bisa digunakan untuk membuat berbagai program berbeda, bukan khusus untuk masalah tertentu saja. Python adalah bahasa pemrograman interpretatif, yaitu instruksi yang terdapat dalam sebuah script dibaca baris demi baris dan langsung dieksekusi. Program yang dikhususkan untuk interpretasi dan eksekusi perintah adalah Python Interpreter. Python dirancang dengan mempertimbangkan keterbacaan. Setiap perancang harus menyadari betapa sulitnya memahami konsep tersebut, tetapi secara umum dapat dikatakan bahwa Python dirancang dengan satu asumsi sederhana yaitu kode dieksekusi oleh mesin, tetapi dibaca oleh manusia. 

Contoh Penggunaan Python 
Python biasa dipakai dalam pengembangan situs dan perangkat lunak, yaitu
1. Data Analisis dan Machine Learning
2. Pengembangan Web (Web Development)
3. Otomatissi Tugas
4. Menyelesaikan Berbagai Tugas Harian
Karena sifatnya yang relatif mudah dipelajari, bahasa pemrograman ini digunakan secara luas oleh non-programmer seperti ilmuwan dan akuntan untuk melakukan tugas harian mereka. Misalnya, dalam mengatur keuangan.

Elemen-elemen Python
1. Variabel dan Tipe Data
    Variabel adalag tempat penyimpanan nilai dalam program. Tipe data dalam Python dapatb  ditentukan otomatis saat runtime.
    1. Integer (int) — bilangan bulat
    2. Float — bilangan desimal
    3. String (str) — teks/karakter
    4. Boolean (bool) — nilai logika True atau False
    5. List — kumpulan data yang dapat diubah (mutable)
    6. Tuple — kumpulan data yang tidak dapat diubah (immutable)
    7. Dictionary (dict) — data berpasangan key-value
    8. Set — kumpulan data unik tanpa urutan

2. Operator
    Operator digunakan untuk melakukan operasi pada nilai atau variabel.
    1. Operator Aritmatika (+, -, *, /, %, **, //)
    2. Operator Perbandingan (==, !=, >, <, >=, <=)
    3. Operator Logika (and, or, not)
    4. Operator Penugasan (=, +=, -=, *=)

3. Struktur Kontrol (Control Flow)
    Digunakan untuk mengatur alur eksekusi program:
    1. Percabangan (Conditional) — if, elif, else
    2. Perulangan (Looping) — for dan while
    3. Kontrol perulangan — break, continue, pass

4. Fungsi (Function)
    Blok kode yang dapat digunakan berulang kali untuk menjalankan tugas tertentu
    1. Fungsi bawaan (built-in function) — seperti print(), len(), input()
    2. Fungsi buatan sendiri (user-defined function) — dideklarasikan menggunakan kata kunci def
    3. Fungsi lambda — fungsi anonim tanpa nama, ditulis dalam satu baris menggunakan lambda

5. Struktur Data Lanjutan
    1. List Comprehension (cara ringkas membuat list)
    2. Nested Data Structure (struktur data di dalam struktur data lain, seperti list di dalam dictionary)

6. Pemrograman Berorientasi Objek (OOP)
    Elemen penting dalam paradigma OOP di Python
    1. Class — cetak biru (blueprint) untuk membuat objek
    2. Object — instansi dari sebuah class
    3. Attribute — variabel yang dimiliki oleh sebuah class/object
    4. Method — fungsi yang dimiliki oleh sebuah class
    5. Inheritance — pewarisan sifat dari class induk ke class turunan
    6. Polymorphism — kemampuan objek memiliki banyak bentuk/perilaku
    7. Encapsulation — pembungkusan data dan method dalam satu kesatuan class

7. Modul dan Package
    1. Modul 
        File .py yang berisi kumpulan fungsi/kelas yang dapat diimpor ke file lain menggunakan import
    2. Package 
        Kumpulan beberapa modul yang diorganisasikan dalam satu folder
    3. Library/Framework 
        kumpulan modul siap pakai

8. Exception Handling
    Mekanisme untuk menangani error/kesalahan saat program berjalan
    1. try = blok kode yang berpotensi menghasilkan error
    2. except = blok kode yang dijalankan jika terjadi error
    3. finally = blok kode yang selalu dijalankan, baik terjadi error maupun tidak
    4. raise = digunakan untuk memicu exception secara manual

9. Komentar (Comment)
    Digunakan untuk memberi keterangan pada kode tanpa memengaruhi eksekusi program
    1. Single-line comment menggunakan tanda "#"
    2. Multi-line comment/docstring menggunakan tanda kutip tiga (''' atau """)
