## 2.1 Dasar Teori
Percabangan adalah salah satu struktur kontrol dasar dalam pemrograman yang memungkinkan program mengambil keputusan berdasarkan suatu kondisi. Tanpa percabangan, program hanya akan berjalan lurus dari atas ke bawah tanpa bisa merespons perbedaan nilai input. Di Python, percabangan dibangun dari kata kunci if, elif, else, dan match-case.
1. IfTunggal
    Bentuk paling sederhana dari percabangan. Program hanya mengevaluasi satu kondisi, dan blok kode di bawahnya hanya dijalankan kalau kondisi itu bernilai True. Kalau kondisinya False, program langsung melompat ke baris setelah blok if tanpa melakukan apa-apa.
2. If-else
    Pengembangan dari if tunggal. Di sini ada dua jalur: kalau kondisi benar jalankan blok if, kalau salah jalankan blok else. Keduanya saling eksklusif yaitu hanya satu yang pasti dieksekusi, tidak mungkin dua-duanya jalan atau dua-duanya tidak jalan sama sekali.
3. If-Elif-Else
    Dipakai kalau kemungkinan hasilnya lebih dari dua. elif (singkatan dari else-if) memungkinkan pengecekan kondisi tambahan setelah kondisi pertama gagal. Python akan mengecek dari atas ke bawah secara berurutan, dan begitu ada satu kondisi yang cocok, blok itu dijalankan lalu langsung keluar dari keseluruhan struktur yaiu sisa elif di bawahnya tidak dicek lagi.
4. Multi-If
    multi-if terpisah ditulis sebagai blok-blok if yang berdiri sendiri-sendiri, tidak dirangkai jadi elif. Sehingga, Python akan mengecek semua blok if satu per satu tanpa peduli blok sebelumnya cocok atau tidak.
5. Match-Case
    Nilai suatu variabel dicocokkan terhadap pola-pola yang didefinisikan lewat case. Begitu ketemu pola yang cocok, blok di bawahnya dijalankan dan langsung keluar dari struktur match.

### 2.2.1 Program "if1_2511531018"
Program ini meminta pengguna memasukkan nilai IPK, lalu memeriksa apakah IPK lebih besar dari 2.75. Jika lebih besar, maka program akan mencetak pesa kelulusan beserta nilai IPK yang telah diinputkan. Program hanya menggunakan if tanpa else, maka jika IPK yang dimasukkan tidak lebih dari 2.75, program tidak akan mencetak pesan apapun terkait status kelulusan, program akan langsung mencetak bahwa program telah selesai.

```python
ipk_1018 = float(input("Input IPK Anda = "))

if ipk_1018 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_1018))

print("Program Selesai")
```

**Penjelasan :**

4. print("Program Selesai") jika kondisi tidak bernilai true, maka program akan selesai.

### 2.2.2 Program "if2_1511531018"

### 2.2.3 Program "multi_if1_2511531018"
Program ini merupakan apakah seseorang boleh mengendarai motor berdasarkan dua faktor yaitu umur dan kepemilikan SIM C. Pengguna diminta memasukkan umur dan status kepemilikan SIM C berupa jawaban "y" atau yang lain (tidak). Program menggunakan empat blok if yang berdiri untuk mengecek keempat kemungkinan kombinasi kondisi umur > 17 dengan SIM, umur > 17 tanpa SIM, umur < !& dengan sim, dan umur < 17 tanpa SIM yang dimana masing-masing menghasilkan pesan yang berbeda

### 2.2.4 Program "if_elif_else1_2511531018"

Dengan ini, program akan berhenti mengecek kondisi begitu menemukan satu kondisi yang benar, dan tidak perlu lagi menuliskan kondisi keempat karena else otomatis menangani semua kemungkinan yang tersisa (yaitu umur <17 dan tidak punya SIM).

### 2.2.5 Program "multi_if2_2511531018"
Program ini menghitung total diskon dan total pembayaran belanja berdasarkan tiga syarat, yaitu total belanja Rp1.000.000, status keanggotaan (member), dan validitas kode promo. Karena ketiga syarat ini dicek menggunakan if terpisah (bukan elif), seorang pembeli bisa saja mendapatkan gabungan dari beberapa diskon sekaligus misalnya diskon belanja besar + diskon member + diskon promo, totalnya bisa mencapai 30%. Program ini juga menunjukkan teknik validasi input yang lebih baik dengan .strip().lower() dan operator in untuk mengecek beberapa kemungkinan jawaban ("y" atau "ya") sekaligus.

```python
#input dari user
total_belanja_1018 = float(input("Masukkan total belanja (Rp): "))

#input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_1018 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_1018 = input_member_1018 in ["y", 'ya']

#input status kode promo (mengecek apakah user mengeik 'y' atau 'ya')
input_promo_1018 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_1018 = input_promo_1018 in ["y", "ya"]

total_diskon_persen_1018 = 0

if total_belanja_1018 > 1000000:
    total_diskon_persen_1018 += 10  #Diskon belanja besar

if is_member_1018:
    total_diskon_persen_1018 += 5 #Diskon member

if kode_promo_valid_1018:
    total_diskon_persen_1018 += 15 #Diskon voucher

#Menghitung nominal diskon dan total bayar
nominal_diskon_1018 = total_belanja_1018 * (total_diskon_persen_1018 / 100)
total_bayar_1018 = total_belanja_1018 - nominal_diskon_1018

#output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon    : {total_diskon_persen_1018}% (Rp {nominal_diskon_1018:,.0f})")
print(f"Total Bayar     : Rp {total_bayar_1018:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_1018}%")
```

**Penjelasan :**
1. Menerima input total belanja sebagai bilangan desimal.
2. Menerima input status member, .strip() menghapus spasi di awal atau akhir, .lower() mengubah semua huruf jadi kecil agar tidak sensitif huruf besar atau kecil.
3. Mengecek apakah jawaban pengguna termasuk dalam daftar ["y", "ya"]; hasilnya berupa boolean (True/False) yang disimpan di is_member_1018.
4. Mengecek apakah kode promo dinyatakan valid berdasarkan jawaban pengguna.
5. Inisialisasi variabel penampung total persentase diskon, dimulai dari 0.
6. Jika belanja lebih dari Rp1.000.000, tambahkan 10% ke total diskon. Operator += berarti menambahkan nilai ke variabel yang sudah ada.
7. Jika pengguna adalah member (True), tambahkan 5% diskon tambahan.
8. Jika kode promo valid, tambahkan 15% diskon tambahan. Ketiga blok if ini independen satu sama lain sehingga diskon bisa terakumulasi.
9. Menghitung nilai nominal (dalam Rupiah) dari diskon berdasarkan persentase total yang terkumpul.
10. Menghitung total yang harus dibayar setelah dikurangi nominal diskon.
11. Mencetak persentase dan nominal diskon; format :,.0f menampilkan angka dengan pemisah ribuan dan tanpa angka desimal.
12. Mencetak total pembayaran akhir dengan format yang sama.
13. Mencetak ulang total persentase diskon yang didapat pengguna.

### 2.2.6 Program "match_case1_2511531018"
Program ini menampilkan nama bulan dalam Bahasa Indonesia berdasarkan angka yang dimasukkan pengguna (1–12). Struktur ini berfungsi mirip seperti switch-case pada bahasa pemrograman lain, dan menjadi alternatif yang lebih rapi dibandingkan menuliskan banyak elif untuk mengecek kesamaan nilai secara berurutan. Setiap case mencocokkan nilai bulan_1018 dengan angka tertentu, dan case _ di akhir berfungsi sebagai kondisi "default" yang menangani semua nilai lain yang tidak cocok, misalnya jika pengguna memasukkan angka di luar rentang 1–12.

```python
bulan_1018 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_1018:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**
1. Menerima input angka bulan dari pengguna sebagai bilangan bulat.
2. Memulai struktur pencocokan pola berdasarkan nilai variabel bulan_1018.
3. Jika bulan_1018 bernilai 1, cetak "Januari". Pola yang sama berulang untuk setiap angka 2 sampai 12, masing-masing mencetak nama bulan yang sesuai (Februari hingga Desember).
4. Simbol garis bawah (_) berarti "apa pun selain yang sudah dicocokkan di atas"; ini menangani input yang di luar rentang 1–12, seperti 0, 13, atau angka negatif.

### 2.2.7 Program "nilai_2511531018"
Program ini mengonversi nilai angka yang dimasukkan pengguna menjadi nilai huruf (A, B, C, D, atau E) menggunakan struktur if-elif-else. Logika pengecekan disusun dari kondisi tertinggi ke terendah, di mana program akan mengecek kondisi secara berurutan dari atas dan langsung berhenti begitu menemukan kondisi yang benar. Pendekatan ini penting untuk kasus rentang nilai seperti ini, karena jika urutannya dibalik (dari terendah ke tertinggi), hasilnya bisa salah — misalnya nilai 90 bisa saja langsung tertangkap oleh kondisi >= 50 jika itu dicek lebih dulu.

```python
nilai_1018 = int(input("Input nilai angka= "))

if nilai_1018 >= 81:
    print("A")
elif nilai_1018 >= 70:
    print("B")
elif nilai_1018 >= 60:
    print("C")
elif nilai_1018 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan :**
1. Menerima input nilai angka dari pengguna sebagai bilangan bulat.
2. Mengecek apakah nilai ≥ 81 (kondisi tertinggi dicek pertama kali). Jika nilai ≥ 81, cetak "A".
3. Jika kondisi sebelumnya salah (berarti nilai < 81), cek apakah nilai ≥ 70 (rentang 70–80). Jika nilai antara 70 dan 80, cetak "B".
4. Jika dua kondisi sebelumnya salah (nilai < 70), cek apakah nilai ≥ 60 (rentang 60–69). Jika nilai antara 60 dan 69, cetak "C".
5. Jika kondisi sebelumnya salah (nilai < 60), cek apakah nilai ≥ 50 (rentang 50–59). Jika nilai antara 50 dan 59, cetak "D".
6. Menangani semua sisa kemungkinan, yaitu nilai < 50. Dicetak jika nilai di bawah 50.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. INPUT DATA PENGUNJUNG & STRING HANDLING
nama_1018 = input("Masukkan Nama Pengunjung        : ")
umur_1018 = int(input("Input umur anda                 : "))
sim_1018 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].strip().lower()

print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")
paket_1018 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_1018 = int(input("Masukkan jumlah tiket           : "))

input_member_1018 = input("Apakah Anda member? (y/t)       : ").strip().lower()
is_member_1018 = input_member_1018 in ["y", "ya"]

input_promo_1018 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()
kode_promo_valid_1018 = input_promo_1018 in ["y", "ya"]

# 2. IF TUNGGAL - VALIDASI KELOGISAN JUMLAH TIKET
if jumlah_tiket_1018 <= 0:
    print("\nPeringatan: Jumlah tiket tidak valid! Transaksi dihentikan.")
    exit()

# 3. MATCH-CASE - PEMILIHAN WAHANA (1-5 DAN DEFAULT _)
match paket_1018:
    case 1:
        nama_wahana_1018 = "Wahana Safari Rimba"
        harga_satuan_1018 = 50000
    case 2:
        nama_wahana_1018 = "Wahana Arung Jeram"
        harga_satuan_1018 = 75000
    case 3:
        nama_wahana_1018 = "Wahana Motor ATV Ekstrim"
        harga_satuan_1018 = 120000
    case 4:
        nama_wahana_1018 = "Wahana Roller Coaster Kilat"
        harga_satuan_1018 = 100000
    case 5:
        nama_wahana_1018 = "Wahana All-Access VIP"
        harga_satuan_1018 = 220000
    case _:
        print("\nPaket wahana tidak valid! Transaksi dihentikan.")
        exit()

# 4. IF-ELIF-ELSE DENGAN OPERATOR LOGIKA (KHUSUS PAKET 3 = ATV)
#    UNTUK PAKET LAIN: IF-ELSE SEDERHANA CEK UMUR >= 10
print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_1018 == 3 and umur_1018 >= 17 and sim_1018 == 'y':
    print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
elif paket_1018 == 3 and umur_1018 >= 17 and sim_1018 != 'y':
    print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
elif paket_1018 == 3 and umur_1018 < 17 and sim_1018 == 'y':
    print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
elif paket_1018 == 3:
    print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
elif umur_1018 >= 10:
    print("Status Akses: Umur Anda memenuhi syarat minimal untuk menaiki wahana ini.")
else:
    print("Status Akses: Umur Anda belum memenuhi syarat minimal (10 tahun) untuk menaiki wahana ini.")

# 5. MULTI-IF TERPISAH - AKUMULASI DISKON BERTINGKAT
subtotal_1018 = harga_satuan_1018 * jumlah_tiket_1018
total_diskon_persen_1018 = 0

if subtotal_1018 >= 200000:
    total_diskon_persen_1018 += 10  # Diskon Belanja Besar

if is_member_1018:
    total_diskon_persen_1018 += 5  # Diskon Member

if kode_promo_valid_1018:
    total_diskon_persen_1018 += 15  # Diskon Voucher Promo

if jumlah_tiket_1018 >= 5:
    total_diskon_persen_1018 += 5  # Diskon Tambahan Rombongan

nominal_diskon_1018 = subtotal_1018 * (total_diskon_persen_1018 / 100)
total_bayar_1018 = subtotal_1018 - nominal_diskon_1018

# 6. IF-ELSE - EVALUASI KELULUSAN AUDIT (BONUS SOUVENIR)
if total_bayar_1018 > 300000:
    catatan_layanan_1018 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_1018 = "Terima kasih telah berkunjung."

# 7. OUTPUT RINCIAN PEMBAYARAN
print("\n--- Rincian Pembayaran ---")
print(f"Nama Pengunjung  : {nama_1018}")
print(f"Wahana Dipilih   : {nama_wahana_1018}")
print(f"Subtotal Belanja : Rp {subtotal_1018:,.0f}")
print(f"Total Diskon     : {total_diskon_persen_1018}% (Rp {nominal_diskon_1018:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_1018:,.0f}")
print(f"Catatan Layanan  : {catatan_layanan_1018}")
print("Program Selesai")
```
